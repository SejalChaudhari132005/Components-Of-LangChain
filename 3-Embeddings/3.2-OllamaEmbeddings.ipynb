```python
from langchain_ollama import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="gemma:2b") #by default, it will use llama2
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Alpha is first character of greek alphabet",
        "Beta is second character of greek alphabet",
    ]
)
```

```python
r1
```

```python
r2=embeddings.embed_query("What is second letter of greek alphabet?")
r2
```

```python
### Other Embedding models
embeddings = OllamaEmbeddings(model="mxbai-embed-large")
text = "This is test document."
query_result = embeddings.embed_query(text)
query_result
```
