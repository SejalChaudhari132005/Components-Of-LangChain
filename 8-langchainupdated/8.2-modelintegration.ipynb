### Streaming and Batch

Stream response while output is been generated

```python
model.invoke("Write 50 words paragraph on Artificial Intelligence.")
```

```python
for chunk in model.stream("Why do parrots have green feathers?"):
    print(chunk.text, end="", flush=True)
```

### Batch

A collection of independent requests to a model which can significantly improve performance and reduce costs as processing can be done parallely

```python
responses= model.batch([
    "Why do parrots talk?",
    "Why is sky blue?",
    "What is quantum computing?"
])

for response in responses:
    print(response.content)
```

```python
model.batch([
    "Why do parrots talk?",
    "Why is sky blue?",
    "What is quantum computing?"], 
config={"max_concurrency":2}
)

```
